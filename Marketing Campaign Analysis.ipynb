### Data Analysis

```python
# Age group
conditions = [
    df["age"] < 30,
    df["age"].between(30, 45),
    df["age"] > 45
]

choices = [
    "Below 30",
    "30-45",
    "Above 45"
]

df["AgeGroup"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df[["age", "AgeGroup"]].head(10))
```

```python
# convert duration in minutes
df["DurationMinutes"] = np.round(
    df["duration"] / 60, 2
)

print(df[["duration", "DurationMinutes"]].head())
```

```python
# Cudtomer Balance group
conditions = [
    df["accountBalance"] < 0,
    df["accountBalance"].between(0, 1000),
    df["accountBalance"].between(1001, 5000),
    df["accountBalance"] > 5000
]

choices = [
    "Negative Balance",
    "0-1000",
    "1001-5000",
    "Above 5000"
]

df["BalanceGroup"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

print(df["BalanceGroup"].value_counts())
```

```python
# mean , median , percentile , standard deviation
balance = df["accountBalance"].dropna().to_numpy()

print("Mean:", np.mean(balance))
print("Median:", np.median(balance))
print("25th Percentile:", np.percentile(balance, 25))
print("75th Percentile:", np.percentile(balance, 75))
print("Standard Deviation:", np.std(balance))
```

```python
# Job analyis
job_analysis = df.groupby("job").agg(
    Total_Customers=("id", "count"),
    Average_Age=("age", "mean"),
    Average_Balance=("accountBalance", "mean"),
    Average_Duration=("duration", "mean")
).round(2)

print(job_analysis)
```

```python
# monthly analysis
month_order = [
    "jan", "feb", "mar", "apr", "may", "jun",
    "jul", "aug", "sep", "oct", "nov", "dec"
]

monthly_analysis = df.groupby("month").agg(
    Total_Records=("id", "count"),
    Average_Duration=("duration", "mean"),
    Average_Balance=("accountBalance", "mean")
).reindex(month_order).round(2)

print(monthly_analysis)

```

```python
# Result Analysis
campaign_analysis = df.groupby(
    "lastCampaignResult"
).agg(
    Total_Records=("id", "count"),
    Average_Duration=("duration", "mean"),
    Average_Balance=("accountBalance", "mean"),
    Average_Contacts=("numberOfContacts", "mean")
).round(2)

print(campaign_analysis)
```

```python
# Pivot Table
pivot = pd.pivot_table(
    df,
    index="job",
    columns="lastCampaignResult",
    values="id",
    aggfunc="count",
    fill_value=0
)

print(pivot)
```

```python
# correlation analysis
numeric_cols = [
    "age",
    "duration",
    "accountBalance",
    "numberOfContacts",
    "numberOfContactsLastCampaign"
]

correlation = df[numeric_cols].corr()

print(correlation.round(2))
```

```python
# to check outliers
balance = df["accountBalance"].dropna()

lower = np.percentile(balance, 25)
upper = np.percentile(balance, 75)

iqr = upper - lower

lower_limit = lower - 1.5 * iqr
upper_limit = upper + 1.5 * iqr

outliers = df[
    (df["accountBalance"] < lower_limit) |
    (df["accountBalance"] > upper_limit)
]

print("Potential outliers:", len(outliers))
print(outliers[["accountBalance", "age", "job"]].head())
```

```python

# analysis columns 
df.to_csv(
    "Marketing_Campaign_Cleaned.csv",
    index=False
)

# To save Analysis tables separatly
job_analysis.to_csv("Job_Analysis.csv")
monthly_analysis.to_csv("Monthly_Analysis.csv")
campaign_analysis.to_csv("Campaign_Analysis.csv")

print("All files exported successfully!")
```
